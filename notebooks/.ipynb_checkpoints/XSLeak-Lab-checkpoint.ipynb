## 2. Victim service:

1. Visit **http://victim.xsleak.test:5000**.
   There is a simple welcome page (HTML/CSS taken from GeeksForGeeks) with some navigation options on the side bar.
2. Browse to `http://victim.xsleak.test:5000/admin.js`. You will recieve a 404 error, as you are not authenicated.
3. Return home and click the `Login` link to set the login cookie. Actual account logic is neglected in this for the sake of focusing  only on the actual XS-leak. 
4. After logging in, you can browse to `http://victim.xsleak.test:5000/admin.js`
   When authenticated, the endpoint will return a simple JavaScript file.

The victim service uses a very simple cookie (`auth=1`) to represent an authenticated user. In a real application, this would be something significantly more secure or complex.
The cookie is set with the `SameSite=Lax` attribute. This can have other values, which will be explored when we add a mitigation later in this lab.



## 3. Attacker service:
The attacker service is a seperate website with malicious code embedded. In real life, these can take the form of traditionally malicious websites, or 

1. Open a new browser tab and navigate to **http://attacker.xsleak.test:8080**.
   You should see a page that explains the lab and includes a **Probe** button.
2. Click the **Probe** button.  The page will create a `<script>` element and attempt to load `/admin.js` from the victim domain.

   - If you are logged in, the resource loads successfully and the `onload` event fires, indicating that the user has access.
   - If you are not logged in, the victim responds with a 404, causing an `error` event to fire instead.

This difference in browser events leaks information across origins, demonstrating an **XS‑Leak** using error events.

## 4. Mitigation strategies

After observing the leak, we must consider various ways to mitigate it:

- **Uniform responses**: We could always return the same code, regardless of the authenticated status. Returning a 404 is a programmer choice.
- **SameSite cookie options**: Use the `SameSite` attribute to disallow authentication cookies to be sent on cross‑site subresource requests.
- **Cross Origin Resource Policy**: Use policies (sent in headers) to prevent the resource from being loaded by cross‑origin pages.

We will go through a series of mitigations. Each time, run the below cell in order to output the current victim_app.py. Then, copy-paste it into the cell below that in order to modify the code. Finally, run the commands after it in order to reset the environment.

```python
file = "../victim/victim_app.py"

with open(file, "r") as f:
    original_code = f.read()

from IPython.display import Code
Code(original_code, language="python")
```

```python
code = """
from flask import Flask, make_response, request, redirect, url_for, render_template
app = Flask(__name__)

AUTH_COOKIE = "auth"
AUTH_COOKIE_VALUE = "1"
SITE_DOMAIN = ".xsleak.test"


@app.route("/")
def index():
    return render_template('base.html')


@app.route("/login")
def login():
    resp = make_response(
        "<p>In theory, this would be a real login page! For demonstration purposes, this is just going to automatically log you in as an 'admin' and give you access to the Admin Panel at <code>/admin.js</code>. "
        "Return to the <a href='/'>home page</a>.</p>"
    )
    resp.set_cookie(
        AUTH_COOKIE,
        AUTH_COOKIE_VALUE,
        domain=SITE_DOMAIN,
        path="/",
        secure=False,
        httponly=True,
        samesite="Strict",
    )
    return resp


@app.route("/logout")
def logout():
    resp = make_response(
        "<p>Logged out. The auth cookie has been cleared. "
        "Return to the <a href='/'>home page</a>.</p>"
    )
    resp.set_cookie(AUTH_COOKIE, "", expires=0, domain=SITE_DOMAIN, path="/")
    return resp

@app.route("/admin.js")
def adminpanel():
    if request.cookies.get(AUTH_COOKIE) == AUTH_COOKIE_VALUE:
        resp = make_response('/* Valid auth, otherwise returns 404 */ window.__victimAuthed = true;', 200)
        resp.headers["Content-Type"] = "application/javascript"
        resp.headers["Cache-Control"] = "no-store"
        return resp
    return "", 200


if __name__ == "__main__":
    app.run(host="0.0.0.0", port=5000)"""

file = "../victim/victim_app.py"

with open(file, "w") as f:
    f.write(code)

print("Code saved.")

```

### B. SameSite Cookie
The SameSite cookie determines when cookies are passed to other sites. Subdomains (i.e. jupyter.xsleak.test vs xsleak.test) would be the "same site" as opposed to xsleak.test vs github.io, which are two different websites. This cookie determines, in instances like these, if the cookie can get passed with requests.

### C. Cross-Origin Resource Policy
Also known as "CORP", this policy controls what type of sources can access the resources of a website. There are multiple values it can be set to, which would potentially allow the same origin vs. the same site vs. anyone to load it. Try enforcing this policy.
